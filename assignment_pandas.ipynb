```python


data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, 21, 24, 22],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE"],
    "Marks": [85, 78, 92, 74, 88]
}

df = pd.DataFrame(data)

print("Complete DataFrame:")
print(df)

print("\nFirst 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))

print("\nShape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

```python
print("Info:")
df.info()

print("\nDescribe:")
print(df.describe())

print("\nNumber of rows:")
print(df.shape[0])

print("\nNumber of columns:")
print(df.shape[1])
```

```python
print("Name column:")
print(df["Name"])

print("\nMarks column:")
print(df["Marks"])

print("\nName and Marks:")
print(df[["Name", "Marks"]])
```

```python
print("First row:")
print(df.iloc[0])

print("\nThird row:")
print(df.iloc[2])

print("\nRows 1 to 3:")
print(df.iloc[1:4])

print("\nName and Marks using loc:")
print(df.loc[:, ["Name", "Marks"]])
```

```python
print("Students scoring above 80:")
print(df[df["Marks"] > 80])

print("\nStudents scoring between 75 and 90:")
print(df[(df["Marks"] >= 75) & (df["Marks"] <= 90)])

print("\nStudents belonging to CSE:")
print(df[df["Department"] == "CSE"])

print("\nCSE students scoring above 80:")
print(df[(df["Department"] == "CSE") & (df["Marks"] > 80)])
```

```python
print("Students belonging to either CSE or ECE:")
print(df[df["Department"].isin(["CSE", "ECE"])])
```

```python
df["Result"] = np.where(df["Marks"] >= 75, "Pass", "Fail")

print(df)
```

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(grade)

print(df)
```

```python
print("Marks in ascending order:")
print(df.sort_values(by="Marks"))

print("\nMarks in descending order:")
print(df.sort_values(by="Marks", ascending=False))

print("\nSorted by Department and then Marks:")
print(df.sort_values(by=["Department", "Marks"]))
```

```python
data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

print("Missing values:")
print(students.isnull())

print("\nNumber of missing values in each column:")
print(students.isnull().sum())

print("\nRows containing missing values:")
print(students[students.isnull().any(axis=1)])
```

```python
students["Age"] = students["Age"].fillna(students["Age"].mean())
students["Marks"] = students["Marks"].fillna(students["Marks"].mean())

print(students)
```

```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],
    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],
    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)

print("Average marks by department:")
print(df.groupby("Department")["Marks"].mean())

print("\nMaximum marks by department:")
print(df.groupby("Department")["Marks"].max())

print("\nNumber of students in each department:")
print(df.groupby("Department")["Name"].count())

print("\nMean, Minimum, Maximum and Count by department:")
print(df.groupby("Department")["Marks"].agg(["mean", "min", "max", "count"]))
```

```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],
    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],
    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],
    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],
    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)

print("First five records:")
print(df.head())

print("\nDimensions:")
print(df.shape)

print("\nAverage Final Mark:")
print(df["FinalMarks"].mean())

print("\nHighest-scoring student:")
print(df.loc[df["FinalMarks"].idxmax()])

print("\nLowest-scoring student:")
print(df.loc[df["FinalMarks"].idxmin()])

print("\nStudents scoring above 80:")
print(df[df["FinalMarks"] > 80])

print("\nAverage marks by department:")
print(df.groupby("Department")["FinalMarks"].mean())

print("\nStudents having attendance above 80%:")
print(df[df["Attendance"] > 80])

print("\nStudents sorted by FinalMarks:")
print(df.sort_values(by="FinalMarks"))

df["Result"] = np.where(df["FinalMarks"] >= 70, "Pass", "Fail")

def performance(mark):
    if mark >= 90:
        return "Excellent"
    elif mark >= 80:
        return "Very Good"
    elif mark >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance"] = df["FinalMarks"].apply(performance)

print("\nDataFrame with Result and Performance:")
print(df)

department_average = df.groupby("Department")["FinalMarks"].mean()

print("\nDepartment having highest average FinalMarks:")
print(department_average.idxmax())
```

```python
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],
    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],
    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

print("Average salary:")
print(employees["Salary"].mean())

print("\nMaximum salary:")
print(employees["Salary"].max())

print("\nMinimum salary:")
print(employees["Salary"].min())

print("\nEmployee receiving the highest salary:")
print(employees.loc[employees["Salary"].idxmax()])

print("\nEmployees earning above ₹60,000:")
print(employees[employees["Salary"] > 60000])

print("\nAverage salary for each department:")
print(employees.groupby("Department")["Salary"].mean())

print("\nAverage experience by department:")
print(employees.groupby("Department")["Experience"].mean())

print("\nNumber of employees in each department:")
print(employees.groupby("Department")["Employee"].count())

print("\nEmployees sorted by salary:")
print(employees.sort_values(by="Salary"))

print("\nIT employees with more than 3 years of experience:")
print(employees[
    (employees["Department"] == "IT") &
    (employees["Experience"] > 3)
])

def salary_level(salary):
    if salary >= 75000:
        return "High"
    elif salary >= 50000:
        return "Medium"
    else:
        return "Low"

employees["SalaryLevel"] = employees["Salary"].apply(salary_level)

print("\nDataFrame with SalaryLevel:")
print(employees)
```
