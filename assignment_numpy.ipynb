```python
# 1D array containing 10 zeros
zeros = np.zeros(10)

# 3 x 3 array containing ones
ones = np.ones((3, 3))

# 4 x 4 identity matrix
identity = np.eye(4)

# Numbers from 5 to 50 with step size 5
numbers = np.arange(5, 51, 5)

print("10 zeros:")
print(zeros)

print("\n3 x 3 ones:")
print(ones)

print("\n4 x 4 identity matrix:")
print(identity)

print("\nNumbers from 5 to 50:")
print(numbers)
```

```python
# Create array containing numbers from 1 to 24
a = np.arange(1, 25)

# Reshape into 4 x 6
a_4x6 = a.reshape(4, 6)

# Reshape into 6 x 4
a_6x4 = a.reshape(6, 4)

# Reshape into 2 x 3 x 4
a_2x3x4 = a.reshape(2, 3, 4)

print("4 x 6:")
print(a_4x6)
print("Shape:", a_4x6.shape)
print("ndim:", a_4x6.ndim)

print("\n6 x 4:")
print(a_6x4)
print("Shape:", a_6x4.shape)
print("ndim:", a_6x4.ndim)

print("\n2 x 3 x 4:")
print(a_2x3x4)
print("Shape:", a_2x3x4.shape)
print("ndim:", a_2x3x4.ndim)
```

```python
salary = np.array([
    25000, 32000, 28000, 45000,
    50000, 38000, 42000, 30000
])

print("Total salary:")
print(np.sum(salary))

print("\nAverage salary:")
print(np.mean(salary))

print("\nHighest salary:")
print(np.max(salary))

print("\nLowest salary:")
print(np.min(salary))

print("\nMedian salary:")
print(np.median(salary))

print("\nStandard deviation:")
print(np.std(salary))
```

```python
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

print("Marks greater than 75:")
print(marks[marks > 75])

print("\nMarks between 50 and 80:")
print(marks[(marks >= 50) & (marks <= 80)])

print("\nNumber of students who scored below 50:")
print(np.sum(marks < 50))

print("\nPercentage of students who scored 60 or above:")
percentage = np.mean(marks >= 60) * 100
print(percentage)
```

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

print("Total marks of each student:")
print(np.sum(marks, axis=1))

print("\nAverage marks of each student:")
print(np.mean(marks, axis=1))

print("\nAverage marks in each subject:")
print(np.mean(marks, axis=0))

print("\nStudent with the highest total:")
totals = np.sum(marks, axis=1)
print(np.argmax(totals) + 1)

print("\nHighest mark in each subject:")
print(np.max(marks, axis=0))
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix addition:")
print(A + B)

print("\nElement-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)

print("\nTranspose of A:")
print(A.T)
```

```python
# A * B performs element-wise multiplication.
# A @ B performs matrix multiplication.

print("A * B:")
print(A * B)

print("\nA @ B:")
print(A @ B)
```

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

average = np.mean(temperature)

print("Mean temperature:")
print(average)

print("\nMaximum temperature:")
print(np.max(temperature))

print("\nMinimum temperature:")
print(np.min(temperature))

print("\nNumber of days above average:")
print(np.sum(temperature > average))

print("\nTemperatures greater than 34°C:")
print(temperature[temperature > 34])

print("\nDifference between maximum and minimum:")
print(np.max(temperature) - np.min(temperature))

print("\nStandard deviation:")
print(np.std(temperature))
```

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Total annual sales:")
print(np.sum(sales))

print("\nAverage monthly sales:")
print(np.mean(sales))

print("\nMonth with highest sales:")
print(np.argmax(sales) + 1)

print("\nMonth with lowest sales:")
print(np.argmin(sales) + 1)

print("\nNumber of months with sales above ₹20,000:")
print(np.sum(sales > 20000))

print("\nPercentage increase from January to December:")
increase = ((sales[-1] - sales[0]) / sales[0]) * 100
print(increase, "%")
```

```python
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

totals = np.sum(marks, axis=1)
averages = np.mean(marks, axis=1)
subject_average = np.mean(marks, axis=0)

print("Total marks of every student:")
print(totals)

print("\nAverage of every student:")
print(averages)

print("\nAverage mark for every subject:")
print(subject_average)

print("\nHighest-scoring student:")
print(np.argmax(totals) + 1)

print("\nLowest-scoring student:")
print(np.argmin(totals) + 1)

print("\nStudents whose average is greater than 80:")
print(np.where(averages > 80)[0] + 1)

print("\nOverall class average:")
print(np.mean(marks))
```

```python
# Generate 20 random integers between 1 and 100
random_numbers = np.random.randint(1, 101, 20)

print(random_numbers)
```

```python
# Find all even numbers
even_numbers = random_numbers[random_numbers % 2 == 0]

print("Even numbers:")
print(even_numbers)
```

```python
# Replace values below 50 with 0
result = np.where(random_numbers < 50, 0, random_numbers)

print("Original array:")
print(random_numbers)

print("\nAfter replacing values below 50:")
print(result)
```

```python
print("Ascending order:")
print(np.sort(random_numbers))

print("\nDescending order:")
print(np.sort(random_numbers)[::-1])
```

```python
unique_values, frequencies = np.unique(
    random_numbers,
    return_counts=True
)

print("Unique values:")
print(unique_values)

print("\nFrequencies:")
print(frequencies)
```

```python
# Example array
a = np.array([10, 20, 30, 40, 50])

# Min-Max normalization
normalized = (a - np.min(a)) / (np.max(a) - np.min(a))

print("Original array:")
print(a)

print("\nNormalized array:")
print(normalized)
```

```python
# Example data
a = np.array([10, 20, 30, 40, 50])

# Standardization
standardized = (a - np.mean(a)) / np.std(a)

print("Original array:")
print(a)

print("\nStandardized array:")
print(standardized)
```

```python
import time

# Create 1,000,000 numbers
list1 = list(range(1000000))
list2 = list(range(1000000))

array1 = np.arange(1000000)
array2 = np.arange(1000000)

# Python list multiplication
start = time.time()

list_result = [a * b for a, b in zip(list1, list2)]

list_time = time.time() - start

# NumPy array multiplication
start = time.time()

array_result = array1 * array2

numpy_time = time.time() - start

print("Time using Python list:")
print(list_time, "seconds")

print("\nTime using NumPy array:")
print(numpy_time, "seconds")
```
