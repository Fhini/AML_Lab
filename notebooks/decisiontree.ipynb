```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/Social_Network_Ads.csv")
df
df.head()        # df.head() gives first five records.
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df
X = df.iloc[:, [2, 3]].values
X
```

```python
y = df.iloc[:, -1].values
y
```

```python
# split data into training set and test set

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=45)    # random state is used so that we get consistent result
X_train
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
X_test
```

```python
from sklearn.tree import DecisionTreeClassifier 
classifier = DecisionTreeClassifier(criterion = 'gini', random_state = 0)
classifier.fit(X_train, y_train)
```

```python
y_pred = classifier.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
Accuracy = (cm[0][0] + cm[1][1])/(cm[0][0]+cm[0][1]+cm[1][0]+cm[1][1])
Accuracy
```

```python
from sklearn import metrics 
prediction = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:",prediction *100,'%')
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
new_individual = np.array([[29, 80000]])
new_individual_scaled = sc.transform(new_individual)
prediction = classifier.predict(new_individual_scaled)

if prediction[0] == 1:
    print("The individual is predicted to purchase.")
else:
    print("The individual is predicted not to purchase.")
```
