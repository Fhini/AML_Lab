```python

import numpy as np
import pandas as p
import matplotlib.pyplot as plt
```

```python
Data = p.read_csv('../data/bestsellers.csv')
dataset = Data
dataset
```

```python
dataset.isnull().sum()
```

```python
dataset.duplicated().sum()
```

```python
dataset
x = dataset.iloc[:, 2:6].values
x
```

```python
y = dataset.iloc[:, -1].values
y
```

```python

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 45)
```

```python
x_test
```

```python
from sklearn import tree
#clf = tree.DecisionTreeClassifier(criterion='entropy', random_state=0)
clf_gini = tree.DecisionTreeClassifier(criterion='gini')
clf_gini.fit(x_train, y_train)
```

```python
from sklearn import tree
#clf = tree.DecisionTreeClassifier(criterion='entropy', random_state=0)
clf_entropy = tree.DecisionTreeClassifier(criterion='entropy')
clf_entropy.fit(x_train, y_train)
```

```python
y_pred_gini = clf_gini.predict(x_test)
y_pred_entropy = clf_entropy.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm_gini = confusion_matrix(y_test, y_pred_gini)
cm_gini
```

```python
from sklearn.metrics import confusion_matrix
cm_entropy = confusion_matrix(y_test, y_pred_entropy)
cm_entropy
```

```python
from sklearn.metrics import accuracy_score
accuracy_gini = accuracy_score(y_test, y_pred_gini)*100
accuracy_gini
```

```python
from sklearn.metrics import accuracy_score
accuracy_entropy = accuracy_score(y_test, y_pred_entropy)*100
accuracy_entropy
```
